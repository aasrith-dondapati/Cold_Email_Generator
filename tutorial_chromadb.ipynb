```python
import chromadb

client = chromadb.Client()
collection = client.create_collection(name="my_collection")
```

```python
collection.add(
    documents = [
        'This document is about Australia',
        'This document is about India'
    ],
    ids = ['id1', 'id2']
)
```

```python
all_docs = collection.get()
all_docs
```

```python
documents = collection.get(ids=['id1'])
documents
```

```python
collection.query(
    query_texts = ['Query is about Sydney'],
    n_results=2
)
```

```python
collection.query(
    query_texts = ['Query is about Hyderabad'],
    n_results=2
)
```

```python
collection.delete(ids=all_docs['ids'])
collection.get()
```

```python
collection.add(
    documents = [
        'This document is about Australia',
        'This document is about India'
    ],
    ids = ['id1', 'id2'],
    metadatas=[
        {"url": "https://en.wikipedia.org/wiki/Australia"},
        {"url": "https://en.wikipedia.org/wiki/India"}
    ]
)
```

```python
collection.query(
    query_texts = ['Query is about Hyderabad'],
    n_results=2
)
```
